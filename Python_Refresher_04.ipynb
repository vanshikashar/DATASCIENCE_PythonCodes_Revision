```python
#JSON module in python
#json.dumps()- convert python object to json string

import json
data= {"name": "Vanshika", "age": 21, "city": "Greater Noida"}

json_string= json.dumps(data)
print(json_string)
print(type(json_string))
```

```python
#converting json to python objects(Deserialization)
#json.loads()- convert json string to python object

json_data= '{"name": "Prisha", "age": 10, "city": "New York"}'
python_obj= json.loads(json_data)
print(python_obj)
print(type(python_obj))
```

```python
#json.load()- read json data from a file

with open("data.json", "r")as file:
    python_data= json.load(file)
print(python_data)
```

```python
#formatting json output

formatted_json= json.dumps(data, indent=4)
print(formatted_json)
```

```python
#object oriented programming in python
#defining a class and creating an object
#creating a class

class Car:
    def __init__(self, brand, model):
        self.brand= brand
        self.model= model

    def display_info(self):
        return f"{self.brand} {self.model}"

car1= Car("BMW", "M4")
print(car1.display_info())
```

```python
#Encapsulation(Data Hiding)

class bankAccount:
    def __init__(self, balance):
        self.__balance= balance

    def get_balance(self):
        return self.__balance

    def deposit(self, amount):
        if amount> 0:
            self.__balance += amount

account= bankAccount(2000)
account.deposit(200)
print(account.get_balance())
```

```python
#Inheritance(Reusing Code)
#Example of single inheritance

class Animal:
    def speak(self):
        return "Animal makes a sound."

class Dog(Animal):
    def speak(self):
        return "Barking"

dog= Dog()
print(dog.speak())
```

```python
#Multiple Inheritance

class A:
    def method_a(self):
        return "Method A"

class B:
    def method_b(self):
        return "Method B"

class C(A, B):
    pass

obj= C()
print(obj.method_a())
print(obj.method_b())
```

```python
#Polymorphism(Same method, different behavior)
#Method Overriding Example

class Bird:
    def fly(self):
        return "Birds can fly"

class Penguin(Bird):
    def fly(self):
        return "Penguins cannot fly"

bird= Bird()
penguin= Penguin()

print(bird.fly())
print(penguin.fly())
```

```python
#Abstraction(Hiding Implementation Details)

from abc import ABC, abstractmethod

class Shape(ABC):
    def area(self):
        pass

class Square(Shape):
    def __init__(self, side):
        self.side= side

    def area(self):
        return self.side* self.side

square= Square(4)
print(square.area())
```

```python
#Magic Methods (Dunder Methods)
#Example- __str__() and __len__()

class Book:
    def __init__(self, title, pages):
        self.title= title
        self.pages= pages

    def __str__(self):
        return f"Book: {self.title}"

    def __len__(self):
        return self.pages

book= Book("Python Programming", 400)
print(str(book))
print(len(book))
```

```python
# class vs static methods
# example

class Example:
    class_var = "I am a class variable"
    
    def instance_method(self):
        return "Instance Method"

    @classmethod
    def class_method(cls):
        return cls.class_var

    @staticmethod
    def static_method():
        return "Static Method"
        
obj = Example()

print(obj.instance_method()) 
print(Example.class_method()) 
print(Example.static_method())
```
