```python
#List Comprehension
#Basic Syntax
#Creating a list of squares
sq= [x**2 for x in range(5)]
print(sq)
```

```python
#using if condition in list comprehension
#Example: Filtering even numbers
evens= [x for x in range(10) if x%2 ==0]
print(evens)
```

```python
#using if-else condition in list comprehension
#Example: Replacing even numbers with "Even" and odd numbers with "odd"
num= ["Even" if x%2== 0 else "Odd" for x in range(5)]
print(num)
```

```python
#Nested loops in list comprehension
#Example: Creating pairs from from two lists
pairs= [(x,y) for x in range(2) for y in range(3)]
print(pairs)
```

```python
#List comprehension with functions
#Example: converting a list of strings to uppercase
words= ["hello", "world", "python"]
upper_words= [word.upper() for word in words]
print(upper_words)
```

```python
#List comprehnsion with nested list comprehension
#Example: FLattening a 2D list
matrix= [[1, 2], [3, 4], [5, 6]]
flattened= [num for row in matrix for num in row]
print(flattened)
```

```python
#List comprehension with set and dictionary comprehension
#Set comprehension
unique_num= {x for x in [1, 2, 2, 3, 4, 4]}
print(unique_num)
```

```python
#Dictionary comprehension
squared_dict= {x: x**2 for x in range(5)}
print(squared_dict)
```

```python
#Performance comparison: List comprehension vs loop
import time
#using a for loop
start= time.time()
squares_loop= []
for x in range(10**6):
    squares_loop.append(x**2)
print("Loop time:", time.time()-start)

#using list comprehension
start= time.time()
squares_comp= [x**2 for x in range(10**6)]
print("List Comprehension time:", time.time()- start)
```

```python
#Lambda Functions
#Example: Simple Lambda function
sq= lambda x: x**2
print(sq(5))
```

```python
#Using lambda functions with map(), filter(), and reduce()
#Using map() with lambda
num= [1, 2, 3, 4]
sq= list(map(lambda x: x**2, num))
print(sq)
```

```python
#Using filter() with lambda
num= [1, 2, 3, 4, 5, 6]
evens= list(filter(lambda x: x%2== 0, num))
print(evens)
```

```python
#Using reduce() with lambda
from functools import reduce
num= [1, 2, 3, 4]
product= reduce(lambda x, y: x*y, num)
print(product)
```

```python
#Lambda with Multiple Arguments
#Example: Adding Two numbers
add= lambda x, y: x+y
print(add(3, 7))
```

```python
#Example: Finding the maximum of two numbers
max= lambda x, y: x if x> y else y
print(max(10, 5))
```

```python
#Lambda in sorting functions
#Sorting a list of tuples
students= [("Anita", 86), ("Babita", 79), ("Charlie", 93)]
students.sort(key= lambda student: student[1])
print(students)
```
